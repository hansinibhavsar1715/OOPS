CLASS IN PYTHON - Addition


With Class

```python
class Addition:
  def __init__(self,a,b):
    self.a=a
    self.b=b
  def add(self):
    return self.a + self.b
obj=Addition(10,20)
result = obj.add()
print("Addition = ",result)
```

Without class

```python
a=10
b=20
sum=a+b
print("Addition=",sum)
```

Without class- simple function version

```python
def add(a,b):
  return a+b
result = add(5,7)
print("Addition = ",result)
```
