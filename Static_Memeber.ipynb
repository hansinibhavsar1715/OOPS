Static Variable

```python
class Student:
  collage = "ABC Collage"       # static variable
  def __init__(self, name):
    self.name = name            # instance variable
  def show(self):
    print("Name:", self.name)
    print("Collage:", Student.collage)
s1 = Student ("Hansini")
s2 = Student ("Hani")
s1.show()
s2.show()
```

Static Method

```python
class Math:
  @staticmethod
  def add(a,b):
    return a+b
result = Math.add(10,20)
print("Addition =", result)
```
