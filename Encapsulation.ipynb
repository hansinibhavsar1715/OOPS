```python
class BankAccount:
  def __init__(self, balance):
    self.__balance = balance      # private variable
  def deposite(self, amount):
    self.__balance +=amount
  def show_balance(self):
    print("Balance:", self.__balance)
acc = BankAccount(1000)
acc.deposite(500)
acc.show_balance()

```

Note: 3 types of Access Modifier

*   public : self.name
*   protected : self._age
*   private : self.__balance




```python
class Student:
  def __init__(self):
    self.__marks = 90
  def get_marks(self):
    return self.__marks
s = Student()
print(s.get_marks())
```
