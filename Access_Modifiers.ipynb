Public Access Modifier

```python
class Student:
  def __init__(self):
    self.name = "Rahul"
s = Student()
print(s.name)
```

Protected Access Modifier

```python
class Student:
  def __init__(self):
    self._marks = 85
class Result(Student):
  def show(self):
    print(self._marks)
r = Result()
r.show()
```

Praivate Access Modifier

```python
class Student:
  def __init__(self):
    self.__age = 20
  def show(self):
    print(self.__age)
s = Student()
s.show()
```

Sealed Modifier

Syntax

```python
class SealedClass:
  def __init_subclass_(cls, **kwargs):
    raise TypeError("This class cannot be inherited")
class child(SealedClass):
  pass

```

Example

```python
class parent:
  def show(self):
    print("Do not override this method")
class child:
  def show(self):
    print("Overriden")
c = child()
#c = parent()
c.show()
```
