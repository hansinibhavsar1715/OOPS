Local Scope

```python
def my_func():
  x=10
  print(x)
my_func()
```

Global Scope

```python
x=20
def my_func():
  print(x)
my_func()
```

Global Keyword

```python
x=10
def my_func():
  global x
  x=50
  print(x)
my_func()
print(x)
```

Enclosing Scope

```python
def outer():
  x=30
  def inner():
    print(x)
  inner()
outer()
```

Nonlocal Keyword

```python
def outer():
  x=10
  def inner():
    nonlocal x
    x=40
    print("Inner x:", x)
  inner()
  print("Outer x:",x)
outer()
```
