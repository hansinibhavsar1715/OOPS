```python
class Student:
  def __init__(self, name, age, grade):   #Constructor
    self.name=name
    self.age=age
    self.grade=grade
  def show(self):
    print("Name:", self.name)
    print("Age:", self.age)
    print("Grade:", self.grade)
s1 = Student("John", 17, 98)         # Object 1
s2 = Student("Krish", 22, 87)        # Object 2
s3 = Student("Sara", 15, 84)         # Object 3
s1.show()
s2.show()
s3.show()
```

Parameterized Constructor

```python
class Student:
  def __init__(self, name, subject):    # parameterized constructor
    self.name = name
    self.subject = subject
  def show(self):
    print("Name:", self.name)
    print("Subject:", self.subject)
obj1 = Student("Rahul", 85)
obj2 = Student("Anjali", 92)

obj1.show()
obj2.show()
```

Copy Constructor

```python
class Student:
  def __init__(self, name, marks):
    self.name = name
    self.marks = marks
  @classmethod              #Copy Constructor
  def copy(cls, obj):
    return cls(obj.name, obj.marks)
  def show(self):
    print("Name:", self.name)
    print("Marks:", self.marks)
s1 = Student("Rahul", 90)

#s2 = Student(s1)
s2 = Student.copy(s1)              # copy object

s1.show()
s2.show()
```

Copy Constructor using Copy module

```python
import copy                            #copy module imported
class Student:
  def __init__(self, name, marks):
    self.name = name
    self.marks = marks
s1 = Student("Rahul", 89)
#s2 = Student.copy(s1)
s2 = copy.copy(s1)
print(s1.name, s1.marks)
print(s2.name, s2.marks)
```

Shallow Copy

```python
import copy
list1 = [1, 2, [3, 4], 5]
list2 = copy.copy(list1)        #Shallow copy

list2[2][0]=99

print(list1)
print(list2)
```

Deep Copy

```python
import copy
list1 = [1,2,[3,4]]
list2 = copy.deepcopy(list1)              # deep copy

list2[2][0] = 99

print("List1:",list1)
print("List2:",list2)
```
