```python
class demo:
  def __init__(self):
    print("Constructor Called")      # constructor

  def __del__(self):
    print("Destructor Called")       # destructor

obj = demo()                         # object creation
del obj                              # object deletion
```

Destructor without del keyword

```python
class Test:
  def __init__(self):
    print("Object Created")      # constructor

  def __del__(self):
    print("Object Destroyed")       # destructor

obj = Test()                         # object creation
print("Program End")
```
