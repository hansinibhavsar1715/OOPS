Abstraction with ABC module

```python
from abc import ABC , abstractmethod

class Animal(ABC):               # abstract base class
    @abstractmethod              # compulsory to implement
    def move(self):
        pass

class Human(Animal):             # first child class
    def move(self):
        print("I can walk and run")

class Snake(Animal):            # second child class
    def move(self):
        print("I can crawl")

H = Human()
H.move()

S = Snake()
S.move()
```

Abstraction without ABC

```python
class Bank:
  def withdraw(self):
    raise NotImplementedError
class ATM(Bank):
  def withdraw(self):
    print("Withdraw cash")
a = ATM()
a.withdraw()
```
